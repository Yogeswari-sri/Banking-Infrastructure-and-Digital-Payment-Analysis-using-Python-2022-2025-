**Task 1**

**Problem Statement:**

The Core Problem: Giant "mega-banks" handle huge transaction volumes. If we calculate simple averages across the entire industry, these giant banks hide the true performance of smaller and regional banks.

Because of this, banking institutions face three main challenges:

**1. Unclear Channel Efficiency**

They don't know which is better: branches, ATMs, or apps.

**2. Lack of Customer Insights**

They don't know what their customers really want.

**3. Poor Asset Allocation**  

They don't know where spend their money in the smart places.

**Business Understanding:**

The main objective of this project is to profile banking performance across commercial and specialized sectors by evaluating customer card adoption, spending efficiency, merchant QR integration, and ATM deployment strategies, and rate banks based on their actual efficiency rather than just their size.

To achieve this, our analysis answers five core business questions:

**Fair Comparison**

How can we compare banks fairly using total active cards instead of just bank size?

**Maintenance**

When is the best time for banks to fix or upgrade their branches and ATMs?

**Real vs. Digital**

What is the right balance between physical branches and online apps?

**Channel Smart Spending**

Which Channel get the most value and profit from the money they spend?

**Bank Market**

How does each bank’s performance compare to the rest of the market?

**Customer Loyalty**

Which banks keep their customers active and using their services regularly?

**Risk & Safety**

Which banks manage financial risk best while growing their digital services?

***Maintainance & Real VS. Digital***

**1. Infrastructure & Deployment Strategy Metrics**

**Onsite vs Offsite Share (%):**

Check where ATMs are placed (Branch vs Public places).

**Transactions per ATM:**

Measures how heavily each ATM machine is used monthly.


```python
# 1. Total Physical ATM Count
df['total_atms'] = (df['atms_crms_onsite'] + df['atms_crms_offsite'])

# 2. Onsite ATM Share percentage % (Branch-Attached Deployment Strategy)
df['onsite_atm_share_pct'] = (((df['atms_crms_onsite'] / df['total_atms']) * 100).replace([np.inf] , np.nan).fillna(0))

# 3. Offsite ATM Share percentage % (Public Accessibility & Reach Strategy)
df['offsite_atm_share_pct'] = (((df['atms_crms_offsite'] / df['total_atms']) * 100).replace([np.inf], np.nan).fillna(0))

# 4. ATM Workload / Stress Level (Informs Maintenance Branch Sector)
atm_total_vol = df["dc_cash_withdraw_atm_vol"] + df["cc_cash_withdraw_atm_vol"]
df["trans_per_atm"] = ((atm_total_vol / df["total_atms"]).replace([np.inf], np.nan).fillna(0))
```

***2. Real vs. Digital***

```python
# Total Digital Infrastructure
df['total_digital_infra'] = df['pos'] + df['micro_atms'] + df['bharat_qr'] + df['upi_qr']
#  Simple Real vs Digital Metric (QR vs Physical POS/ATM Ratio)
# Digital (QR) vs Physical (POS + Micro ATM)
digital_infra = df['bharat_qr'] + df['upi_qr']
physical_infra = df['pos'] + df['micro_atms']

df['digital_vs_real_ratio'] = (digital_infra / physical_infra).replace(np.inf, np.nan).fillna(0)
```

***3. Channel Smart Sprnding & Risk/Safety***


Channel-wise Average Analysis(Channel vs Channel)

Calculates the average transaction value across PoS, Online, Other, and Cash Withdrawal channels by dividing Total Value by Total Volume.



```python
# Debit Card Channel-wise Average Ticket Size (ATV)
# Average spend per PoS transaction (Point of Sale Swiping)
df["dc_pos_avg"] = (df["dc_pay_trns_at_pos_val"] / df["dc_pay_trns_at_pos_vol"]).replace([np.inf], np.nan).fillna(0)
# Average spend per Online transaction (E-commerce / Web)
df["dc_online_avg"] = (df["dc_pay_trns_online_val"] / df["dc_pay_trns_online_vol"]).replace([np.inf], np.nan).fillna(0)

# Credit Card Channel-wise Average Ticket Size (ATV)
# Average spend per PoS transaction
df["cc_pos_avg"] = (df["cc_pay_trns_at_pos_val"] / df["cc_pay_trns_at_pos_vol"]).replace([np.inf], np.nan).fillna(0)
# Average spend per Online transaction
df["cc_online_avg"] = (df["cc_pay_trns_online_val"] / df["cc_pay_trns_online_vol"]).replace([np.inf], np.nan).fillna(0)
```

***4. Fair Comparision & Customer Loyalty (Normalized Metrics per Active Card)***

 Overall Average across Debit and Credit card

Measure how much money a customer spend transactions per card across both cards.

```python
# Total Volumes and Values
dc_tot_val = df["dc_pay_trns_at_pos_val"] + df["dc_pay_trns_online_val"] + df["dc_pay_trns_other_val"] + df["dc_cash_withdraw_atm_val"] + df["dc_cash_withdraw_pos_val"]
dc_tot_vol = df["dc_pay_trns_at_pos_vol"] + df["dc_pay_trns_online_vol"] + df["dc_pay_trns_other_vol"] + df["dc_cash_withdraw_atm_vol"] + df["dc_cash_withdraw_pos_vol"]

cc_tot_val = df["cc_pay_trns_at_pos_val"] + df["cc_pay_trns_online_val"] + df["cc_pay_trns_other_val"] + df["cc_cash_withdraw_atm_val"]
cc_tot_vol = df["cc_pay_trns_at_pos_vol"] + df["cc_pay_trns_online_vol"] + df["cc_pay_trns_other_vol"] + df["cc_cash_withdraw_atm_vol"]

# Spend & Transaction Frequency per Active Card (Fair Comparison & Loyalty)
df["monthly_spend_per_debit_card"] = ((dc_tot_val / df["debit_cards"]).replace([np.inf], np.nan).fillna(0))
df["monthly_spend_per_credit_card"] = ((cc_tot_val / df["credit_cards"]).replace([np.inf], np.nan).fillna(0))
df["monthly_trans_per_debit_card"] = ((dc_tot_vol / df["debit_cards"]).replace([np.inf], np.nan).fillna(0))
df["monthly_trans_per_credit_card"] = ((cc_tot_vol / df["credit_cards"]).replace([np.inf], np.nan).fillna(0))

# Calculate average withdrawal amount per ATM transaction
df["cc_atm_avg_withdrawal_amt"] = (df["cc_cash_withdraw_atm_val"] / df["cc_cash_withdraw_atm_vol"]).replace([np.inf], np.nan).fillna(0)
```

***6. Customer Loyalty***

Which banks keep their customers active and using their services regularly?

This code groups banks by category, calculates average debit and credit card transactions, adds a combined total, and plots a stacked bar chart with a red benchmark line to compare customer loyalty across categories.

```python
# 1. Group by Bank Category and calculate average monthly transactions
loyalty_df = df.groupby('bank_category')[
    ['monthly_trans_per_debit_card', 'monthly_trans_per_credit_card']
].mean().reset_index()

# 2. Add total combined transactions column
loyalty_df['Total_Trans'] = (
    loyalty_df['monthly_trans_per_debit_card'] +
    loyalty_df['monthly_trans_per_credit_card']
)

# Sort from most loyal to least loyal
loyalty_df = loyalty_df.sort_values(by='Total_Trans', ascending=False)

# 3. Create Stacked Bar Chart for Loyalty & Frequency Comparison
plt.figure(figsize=(9, 5))

plt.bar(
    loyalty_df['bank_category'],
    loyalty_df['monthly_trans_per_debit_card'],
    label='Debit Card Trans',
    color='#2b5c8f'
)
plt.bar(
    loyalty_df['bank_category'],
    loyalty_df['monthly_trans_per_credit_card'],
    bottom=loyalty_df['monthly_trans_per_debit_card'],
    label='Credit Card Trans',
    color='#d95f02'
)
# Market Benchmark Line (Combined Total Average)
market_loyalty_avg = (
    df['monthly_trans_per_debit_card'].mean() +
    df['monthly_trans_per_credit_card'].mean()
)
plt.axhline(
    market_loyalty_avg,
    color='red',
    linestyle='--',
    linewidth=2,
    label=f'Market Avg ({market_loyalty_avg:.1f} Trans)'
)

plt.title("Customer Loyalty: Total Monthly Transactions per Card by Bank Category")
plt.xlabel("Bank Category")
plt.ylabel("Combined Monthly Transactions per Card")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

***6. Customer Loyalty***

**Interpretation** The chart shows how active customers are across bank categories, measured by monthly transactions per card compared to the market average (4.6).

**Key Insight** Private Sector Banks lead with the strongest loyalty, Foreign Banks also above average, while Public and Small Finance Banks hover near or below average, and Payment Banks show the weakest loyalty.

**Pattern** Debit + credit activity is balanced in Private and Foreign Banks, debit dominates in Public and Small Finance, and Payment Banks remain inactive.

**Trend** Loyalty is uneven: strong in private/premium banks, moderate in public/small finance, very weak in payment banks.

**Observation** Banks must tailor strategies: reward frequent users (Private/Foreign), push digital adoption (Public/Small Finance), and pivot to mobile-first offers (Payment Banks).
